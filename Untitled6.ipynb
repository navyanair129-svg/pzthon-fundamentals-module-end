```python
feedbacks = [
    "The service was good and fast.",
    "Poor quality, extremely disappointed.",
    "Excellent experience! Highly recommended.",
    "Good value for money.",
    "The food was good, but the service was poor.",
    "Absolutely excellent service."
]
def count_word_in_feedbacks(word):
    count = 0
    target_word = word.lower()
    
    for feedback in feedbacks:
        if target_word in feedback.lower():
            count += 1
            
    return count

print(f"Number of feedbacks containing 'good': {count_word_in_feedbacks('good')}")
print(f"Number of feedbacks containing 'poor': {count_word_in_feedbacks('poor')}")
print(f"Number of feedbacks containing 'excellent': {count_word_in_feedbacks('excellent')}")
```

```python
feedback_data = {
    "rating": [5, 2, 5, 4, 3, 5],
    "comment": [
        "The service was good and fast.",
        "Poor quality, extremely disappointed.",
        "Excellent experience! Highly recommended.",
        "Good value for money.",
        "The food was good, but the service was poor.",
        "Absolutely excellent service."
    ]
}

print("--- Cleaned Feedback Data ---")
print(feedback_data)


# 2.average rating 
average_rating = sum(feedback_data["rating"]) / len(feedback_data["rating"])
print(f"\nAverage Rating: {average_rating:.2f}")

# 3.
longest_comment = max(feedback_data["comment"], key=lambda comment: len(comment.split()))
print(f"\nFeedback with the longest comment ({len(longest_comment.split())} words):")
print(f"\"{longest_comment}\"")

# 4. Print the list of unique words used across all feedbacks (lowercased, ignoring punctuation)
unique_words = set()
for comment in feedback_data["comment"]:
    # Clean up punctuation to prevent words like 'fast.' and 'fast' from being separate entries
    cleaned_comment = comment.translate(str.maketrans("", "", string.punctuation)).lower()
    unique_words.update(cleaned_comment.split())

print(f"\nUnique Words Used ({len(unique_words)} words):")
print(sorted(list(unique_words)))

# 5. 🧩 (Optional): Sort feedbacks by rating (highest to lowest) using zip() and sorted()
print("\n--- Feedbacks Sorted by Rating (Highest to Lowest) ---")
# zip elements together, sort descending based on the rating element (index 0)
sorted_entries = sorted(zip(feedback_data["rating"], feedback_data["comment"]), key=lambda x: x[0], reverse=True)

for rating, comment in sorted_entries:
    print(f"Rating: {rating} ★ | Comment: {comment}")
```
